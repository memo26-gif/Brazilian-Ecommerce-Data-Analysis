# 2. Merging & Cleaning Datasets

In this section, we combine all the isolated datasets into a single, comprehensive DataFrame (`df_merged`) using left joins. This allows us to link customer orders with full product details, aisles, and departments for a deeper analysis.

### Merging Strategy:
* Connect prior order products with overall order details via `order_id`.
* Map product information (names, categories) using `product_id`.
* Link product_category_name_translation and customers_id metadata using `product_category_name_translation` and `customer_id`.

```python
df_merged = pd.merge(order_items_df, orders_df, on='order_id', how='left')
df_merged = pd.merge(df_merged, products_df, on='product_id', how='left')
df_merged = pd.merge(df_merged,products_category_name_translation_df, on='product_category_name', how='left')
df_merged = pd.merge(df_merged, customers_df, on='customer_id', how='left')
df_merged.head(15)
```

# Task 1: Top 10 Buyer States Analysis


```python

top_states = df_mergen_clean['customer_state'].value_counts().head(10)

plt.figure(figsize=(15, 6))

sns.barplot(x=top_states.values, y=top_states.index, palette='Blues_r')

plt.title('Top 10 Buyer States (Most Orders)', fontsize=14, fontweight='bold')
plt.xlabel('Number of Orders', fontsize=12)
plt.ylabel('State Code', fontsize=12)

plt.tight_layout()
plt.show()
```

**Business Recommendation:** The marketing team should allocate at least 50% of the digital ad budget specifically to São Paulo (SP) to maximize the Return on Ad Spend (ROAS), while optimizing supply chain operations in runner-up states like RJ and MG to capture untapped potential.

# Task 3: When Do Customers Shop? (Orders by Hour of the Day)

```python

df_mergen_clean['order_purchase_timestamp'] = pd.to_datetime(df_mergen_clean['order_purchase_timestamp'])
df_mergen_clean['order_hour_of_day'] = df_mergen_clean['order_purchase_timestamp'].dt.hour

hourly_orders = df_mergen_clean['order_hour_of_day'].value_counts().sort_index()

plt.figure(figsize=(15, 6))

sns.barplot(
    x=hourly_orders.index, 
    y=hourly_orders.values, 
    palette='Blues_r', 
    edgecolor='grey'
)

plt.title('Hourly Customer Purchase Patterns', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Hour of the Day (00:00 - 23:00)', fontsize=12, fontweight='bold')
plt.ylabel('Number of Orders', fontsize=12, fontweight='bold')

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Objective:** To analyze customer shopping behaviors across different hours of the day to identify peak ordering windows.

**Key Insight:** Sales volume shows a strong concentration during midday and afternoon hours, peaking specifically between 10:00 AM and 4:00 PM, while dropping drastically during late-night and early-morning hours (2:00 AM - 6:00 AM).

**Business Recommendation:** The digital marketing team should schedule promotional push notifications and major email campaigns right before peak hours (around 9:30 AM) to capture maximum user attention. Furthermore, server maintenance and IT updates should be scheduled during the lowest activity window (3:00 AM to 5:00 AM) to avoid hurting user experience and losing potential sales.


# Task 4: High-Value vs. High-Freight Categories (New Visual Style)


```python
# 1. حساب نسبة تكلفة الشحن من سعر المنتج كعمود جديد
df_merged['Freight_Percentage'] = (df_merged['freight_value'] / df_merged['price']) * 100

# 2. حساب متوسط النسبة وتجميع البيانات لأعلى 5 فئات
top_freight_impact = df_merged.groupby('product_category_name_english').agg(
    Freight_Percentage=('Freight_Percentage', 'mean')
).nlargest(5, 'Freight_Percentage').reset_index()
```

```python

plt.figure(figsize=(15, 6))



sns.pointplot(
    data=top_freight_impact,
    x='product_category_name_english', 
    y='Freight_Percentage',
    color='skyblue',     
    markers='o',           
    linestyles='-'        
)

sns.scatterplot(
    data=top_freight_impact,
    x='product_category_name_english', 
    y='Freight_Percentage',
    hue='Freight_Percentage',
    palette='Blues_r',
    s=200,                 
    legend=False,
    edgecolor='grey',
    zorder=3              
)

plt.title('Top 5 Categories with Highest Shipping Cost Burden', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Product Category', fontsize=12, fontweight='bold')
plt.ylabel('Shipping Cost as % of Product Price', fontsize=12, fontweight='bold')

# plt.xticks(rotation=15, ha='right')

plt.grid(axis='both', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Objective:** To identify the top 5 product categories where shipping costs represent an alarmingly high percentage of the product's retail price, which directly triggers cart abandonment.

**Key Insight:** Low-cost or bulkier items like Home Comfort (home_comfort_2) and Flowers (flowers) suffer from a massive logistics cost burden, where shipping fees devour 45% to over 55% of the item's baseline value. This is driven by volumetric weight pricing and special handling requirements.

**Business Recommendation:** Management must abandon the traditional pricing model for these specific 5 categories. Instead, they should adopt a "Price Bundling / Hidden Shipping" strategy by embedding the shipping fee into the product price and offering "Free Shipping" on the website. Psychologically, this eliminates checkout friction and boosts conversion rates significantly.


# Task 5: Product Weight vs. Shipping Cost Analysis

```python

plt.figure(figsize=(15, 6))

sns.scatterplot(
    data=df_mergen_clean.sample(5000, random_state=42),
    x='product_weight_g',
    y='freight_value',
    color='skyblue',       
    alpha=0.6,            
    edgecolor='grey'      
)

plt.title('Does Product Weight Increase Shipping Cost?', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Product Weight (in Grams)', fontsize=12, fontweight='bold')
plt.ylabel('Shipping Cost (Freight Value)', fontsize=12, fontweight='bold')

plt.grid(axis='both', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Objective:** To investigate the statistical relationship between product weight and actual shipping fees (freight value) to optimize delivery pricing.

**Key Insight:** While there is a general trend that heavier items cost more to ship, the chart reveals significant anomalies: many lightweight products incur exceptionally high shipping fees. This proves that logistics providers are heavily factoring in volumetric weight (package dimensions) rather than just actual weight.

**Business Recommendation:** The operations team should implement automated box-sizing algorithms to reduce dead space in packaging. By minimizing oversized packaging for light items, the company can instantly lower freight expenses, pass the savings to customers, and boost checkout conversions.
